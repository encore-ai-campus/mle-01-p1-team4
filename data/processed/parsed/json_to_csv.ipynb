```python
import json
import csv
from pathlib import Path

input_dir = Path(".")
output_file = Path("..") / "document_inspection.csv"

documents = list(input_dir.glob("*.json"))

rows = []

for document in documents:

    with open(document, "r", encoding="utf-8") as f:
        data = json.load(f)

    for item in data:
        rows.append({
            "문서": document.name,
            "페이지": item["metadata"]["page"],
            "종류": item["metadata"]["category"],
            "내용": item["page_content"]
        })

with open(output_file, "w", encoding="utf-8-sig", newline="") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["문서", "페이지", "종류", "내용"]
    )
    writer.writeheader()
    writer.writerows(rows)

print(f"저장 완료: {output_file.resolve()}")
print(f"문서 수: {len(documents)}")
print(f"전체 요소 수: {len(rows)}")
```
